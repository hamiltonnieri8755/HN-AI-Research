```python
# import packages
import numpy as np
import matplotlib.pyplot as plt
from sklearn.naive_bayes import GaussianNB
from sklearn.model_selection import train_test_split
from sklearn.model_selection import cross_val_score
```

```python
# import file containing data
data = np.loadtxt('content/data_multivar_nb.txt', delimiter=",")
data
```

```python
# get X and y
X = data[:, :-1]
y = data[:, -1]
X.shape
```

```python
# create naive bayes classifier
nb_classifier = GaussianNB()
```

```python
# Train the classifier
nb_classifier.fit(X, y)
```

```python
# predict values for training data
y_pred = nb_classifier.predict(X)
```

```python
accuracy = 100.0 * (y == y_pred).sum() / X.shape[0]
print(f'accuracy is {accuracy}')
```

```python
# split data into training and testing subsets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=3)
```

```python
# create a new Naive Bayes classifier
nb_classifier_new = GaussianNB()
```

```python
# train the new classfier with the split data
nb_classifier_new.fit(X_train, y_train)
```

```python
# predict x_test
y_test_pred = nb_classifier_new.predict(X_test)
```

```python
# compute accuracy of the new classifier
accuracy_new = (y_test == y_test_pred).sum() / X_test.shape[0] * 100.0
print(f'accuracy new is {accuracy_new}')
```

```python
# calculate cross validation score
num_folds = 3
accuracy_cross_val = cross_val_score(nb_classifier_new, X, y, scoring='accuracy', cv=num_folds)
print(f'accuracy_cross_val new {str(round(accuracy_cross_val.mean() * 100.0, 2))}%')
```

```python
accuracy_cross_val_old = cross_val_score(nb_classifier, X, y, scoring='accuracy', cv=3)
print(f'accuracy_cross_val old {str(round(accuracy_cross_val_old.mean() * 100.0, 2))}%')
```
